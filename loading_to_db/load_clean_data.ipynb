Loading data on Postgresql database

```python
import sys
import os
import pandas as pd
from sqlalchemy import text

# Add project root to sys.path
sys.path.append(os.path.abspath(r"C:\Users\kpapa\cars_etl_project"))

# Import the DB engine creator
from loading_to_db.db_connection import create_db_engine

# Create engine
engine = create_db_engine()

# Path to cleaned data
clean_data_path = r"C:\Users\kpapa\cars_etl_project\loading_to_db\Car_sale_ads_clean.csv"

# Read and load data in chunks
df_reader = pd.read_csv(clean_data_path, chunksize=10000)

for chunk in df_reader:
    if "Currency" in chunk.columns:
        chunk.drop(columns=["Currency"], inplace=True)
    chunk.to_sql('cars', engine, if_exists='append', index=False)
    print(f"Loaded chunk with {len(chunk)} rows.")

print("Data loaded successfully!")
```

Creating engine to connect  with the  database

```python
import os
from dotenv import load_dotenv
from sqlalchemy import create_engine


load_dotenv()

DB_NAME = os.getenv("DB_NAME")
DB_USER = os.getenv("DB_USER")
DB_PASS = os.getenv("DB_PASS")
DB_HOST = os.getenv("DB_HOST")
DB_PORT = os.getenv("DB_PORT")


db_url = f"postgresql://{DB_USER}:{DB_PASS}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
engine = create_engine(db_url)
```

Checking for connection with postgresql database

```python
from sqlalchemy import text

with engine.connect() as conn:
    result = conn.execute(text("SELECT version();"))
    print(result.fetchone())
```

Performing a query to test the connection

```python
from sqlalchemy import create_engine, text

# connection with database through SQLAlchemy
db_url = f"postgresql://{DB_USER}:{DB_PASS}@{DB_HOST}:{DB_PORT}/{DB_NAME}"
engine = create_engine(db_url)

# test query
with engine.connect() as conn:
    result = conn.execute(text("SELECT * FROM cars LIMIT 10;"))
    rows = result.fetchall()

# get column names
columns = result.keys()

# print result
for row in rows:
    print(dict(zip(columns, row)))
```
